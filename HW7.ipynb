````python
import pandas as pd
````

````python
data_covid = pd.read_csv(covid_file_path)
data_covid
````

````python
data_covid.head()
````

````python
data_covid['sex'].isnull()
````

````python
missing_sex = data_covid[data_covid['sex'].isnull()]
missing_sex
````

````python
missing_sex.groupby('nationality').describe()
````

````python
missing_sex.groupby('province_of_onset').describe()
````

````python
missing_sex.groupby(['province_of_onset','nationality']).describe() ## groupby หลาย column
````

````python
misssing_sex_no_burma = missing_sex[missing_sex['nationality']!='Burma']
misssing_sex_no_burma.groupby('risk').describe()
````

````python
missing_sex.groupby('risk').describe()
````

````python
missing_sex[missing_sex['risk']=='Cluster สมุทรสาคร'].groupby('announce_date').count()
````

สรุปสาเหตุที่ข้อมูล sex เป็น missing:
ข้อมูลเพศที่สูญหาย (missing) ส่วนใหญ่เกิดจากกรณี Cluster สมุทรสาคร ในช่วงปลายเดือนธันวาคม 2020 ซึ่งเป็นการตรวจคัดกรองเชิงรุกกับกลุ่มแรงงานต่างด้าวจำนวนมากในเวลาอันสั้น ในช่วงเวลาดังกล่าวเน้นการบันทึกข้อมูลเพื่อควบคุมโรคอย่างเร่งด่วน ทำให้การลงประวัติส่วนบุคคล (เช่น เพศ และ อายุ) ไม่สมบูรณ์และถูกเว้นไว้เป็นค่าว่าง
