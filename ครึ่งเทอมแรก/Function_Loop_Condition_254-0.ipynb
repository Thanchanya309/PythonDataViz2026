## Function template
#### ทำหน้าที่รับ input มาประมวณผลออกมาเป็น output

### def คือการกำหนดฟังก์ชัน


backtick (`) ==> กด ~ ค้าง, alt - 9>6 (full keyboard with number)

tilde (~)

curly bracket ({ })

square bracket ([ ])

```python
    def function_name(_Input_) :
        do_something with _Input_ to get _Output_
        return _Output_
```

เขียน def ไว้ก่อนเป็นการบอกว่าจะสร้างฟังก์ชัน จากนั้นตามด้วยชื่อฟังก์ชั่น หลังจากนั้นตามด้วย input เราอยากให้ฟังก์ชันรับค่าของอะไรมา เมื่อรับ input เข้าก็จะประมวลผลเผื่อให้ได้ค่า output โดยส่ง return ออกมา

function มีส่วนสำคัญทั้งหมด 4 ส่วน

1.   บอก python ว่าเราจะเขียนฟังก์ชั่น ชื่ออะไร  ```def function_name():``` (ขาดไม่ได้)
2.   กำหนดตัวแปรที่จะเป็น input  ```_Input_``` (ขาดได้)
3.   ส่วนประมวลผล ```do_something with _Input_ to get _Output_``` (ขาดไม่ได้)
4.   ส่วน output ```return _Output_```(ขาดได้)



- การกด ` หรือ backtick ได้จาก Alt+96
- ~ หรือ tilde ไก้จาก สลับเป็นอังกฤษ แล้วกด ship + ตัวหนอน
- { } หรือ curly bracket
- ( [ ] ) หรือ square bracket

#### ตัวอย่างการเขียน normal function


````python
#ต้องการสร้างฟังก์ชันชื่อว่า print_name มี intput 3 ตัว
def print_name(surname,ID,name):
    st = f'ชื่อ {name} นามสกุล {surname} รหัส {ID}' #โดยเราจะสร้างตัวแปรขึ้นมาใหม่ชื่อว่า st
    return st #มันไม่โชว์ออกมาแต่เก็บไว้ในความจำไว้แล้ว
````

````python
def print_2lines_default(name,surname,ID,grade='F'):
    st = f'ชื่อ {name} นามสกุล {surname} รหัส {ID}'
    print (st)
    st2=f'เกรดวิชา Data Viz >>> {grade}'
    print(st2)
````

# LOOP การวนซ้ำ

for เป็นคำที่ใช้บอก python ว่าเรากำลังเขียน loop
โดย for จะวนดึงสมาชิกจาก ```listA``` มาทำ process ```do_something```

```python
    for each_member in listA :
        do_something
```

#### คำสั่ง for เป็นคำสั่งวนซ้ำที่ใช้ควบคุมการทำงานซ้ำๆ ในจำนวนรอบที่แน่นอน อยากให้มันวนอะไรก็สร้าง list แล้วจับมาให้มันวนใน list ได้เลย

````python
# มี for ต้องมี list ดึวมาตามลำดับ
# for จะไปดึงสมาชิกใน list ตัวแรกมาทำพอทำเสร็จก็ดึตัวต่อไปเรื่อยๆ วนจนหมด
for i in [1,2,3] :
    o = i**2 # เอาแต่ละสมากชิกมายกกำลัง
    print (f'this member = {i} after process = {o}')
````

### loop in loop
#### สามารถใช้ for loop ซ้อนกันสองรอบได้ ซ้อนกี่รอบก็ได้

เริ่มมามันจะวิ่งตั้งแต่ 2 ไปจนถึง 4 ก่อน (เอาถึง 5 แต่ไม่ได้เอา 5) หลังจากวิ่ง 2 ตัวแรกเสร็จ แล้วมันก็จะมาเข้าลูปนี้เลยเพราะลูปนี้อยู่ข้างในมัน โดยความหมายมันก็คือ ตอนนี้ name1 = 2 นะ แล้วเข้ามาในลูป 2 ซึ่งลูปที่สองก็คือ name2 ก็จะวิ่งตั้งแต่ 1 ไปจนถึง 12 (name2 เปลี่ยนไปเรื่อยๆ โดยที่ name 1 มันยังเป็น 2 อยู่) พอหมดก็จะออกมา

````python
for name1 in  range(2,5): # เริ่มมามันจะวิ่งตั้งแต่ 2 ไปจนถึง 4 ก่อน (เอาถึง 5 แต่ไม่ได้เอา 5)
    for name2 in  range(1,13): # หลังจากวิ่ง 2 ตัวแรกเสร็จ แล้วมันก็จะมาเข้าลูปนี้เลยเพราะลูปนี้อยู่ข้างในมัน
        print(name1,' x ',name2,' = ',name1*name2)
    print('the inner loop is end')
````

````python
for name1 in  range(2,5): # เริ่มมามันจะวิ่งตั้งแต่ 2 ไปจนถึง 4 (เอาถึง 5 แต่ไม่ได้เอา 5)
    print(f'now name1 = {name1}') # เพิ่มเข้ามาเพื่อนเป็นตัวบอกว่าตอนนี้ name1 = เท่าไหร่ กันงง
    for name2 in  range(1,13):
        print(name1,' x ',name2,' = ',name1*name2)
    print('the inner loop is end')
````

### loop in function
#### การวนลูปในฟังก์ชั่น¶

````python
list_friends = ['พี่บูม','ดายน์','แตงโม','ฝน','พี่เจ็ท','น้องแคมป์']
````

````python
def print_feelings(I,friendS,feeling = 'คิดถึง'): # input 3 ตัว คือ ชื่อตัวเอง,เพื่อน,ความรู้สึก(ในที่นี้เรากำหนดวไ้ว่า คือ คิดถึง)
    for friend in friendS:
        print(f'{I} {feeling} {friend}')
````

### conditional คือ เงื่อนไข /ประโยคเงื่อนไข

##### True / False
##### การวนลูปโดยมีเงื่อนไขต่างๆมาเกี่ยวข้อง โดยมี pattern ดังนี้


```Python
if condition1 :
    do something
elif condition2 :  #elif คือ else if
    do another thing
else:
    do ...
```
    

````python
print(list_friends)
````

````python
for name1 in list_friends :
    for name2 in list_friends :
            print(name1,'รัก',name2)
````

````python
for name1 in list_friends:
    for name2 in list_friends:
        if name1 != name2: #ไม่ปริ้นชื่อคนเดียวกันซ้ำ
            print(name1,'รัก',name2)
````

````python
for name1 in list_friends:
    for name2 in list_friends:
        if name1 == name2: #ปริ้นชื่อคนเดียวกันซ้ำ
            print(name1,'รัก',name2)
````

````python
for name1 in list_friends:
    for name2 in list_friends:
        if name1 == name2:
            'do nothing' # ไม่สามารถเว้นว่างได้ต้องบอกมันหน่อย
        else:
            print(name1,'รัก',name2)
````
