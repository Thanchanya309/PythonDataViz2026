## ให้สร้างฟังก์ชั่นคำนวณเวลาเป็นวินาทีของเวลาใดๆ และ print ออกมาให้สวยงาม

````python
time = ['12:30:15', '13:41:07', '12:53:15', '00:59:25', '11:11:11', '16:06:09']

def print_sec(time):
    h,m,s = time.split(':')
    sec = int(h) * 3600 + int(m) * 60 + int(s)
    print(f'เวลา {time} คิดเป็น {sec} วินาที')
````

````python
print_sec('12:30:15')
print_sec('13:41:07')
print_sec('12:53:15')
print_sec('00:59:25')
print_sec('11:11:11')
````

## ให้ สร้าง list ของ เวลา ```['12:30:15','13:41:07',....]``` แล้ววนลูปเรียกฟังก์ชั่นคำนวณเวลาเป็น วินาที

**commit ว่า HW3**

````python
# 1. สร้าง ฟังก์ชัน คำนวณเวลาเป็นวินาที
def time_to_sec(time_str):
    h,m,s = time_str.split(':')
    total = int(h) * 3600 + int(m) * 60 + int(s)
    return total

# 2. สร้าง list ของเวลา
times = ['12:30:15','13:41:07','12:53:15','00:59:25','11:11:11','16:06:09']

# 3. วนลูปเรียกใช้ฟังก์ชันคำนวณ
for t in times: # times ไม่ต้องมีอะไรครอบ เพราะ times เป็นตัวแปรประเภท List อยู่แล้ว
    sec = time_to_sec(t)
    print(f'เวลา {t} คิดเป็น {sec} วินาที')
````
