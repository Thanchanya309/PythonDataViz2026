````
this_data
````

````
# อาจาร์ยพาทำ
female = this_data[this_data['sex']=='หญิง']
female.describe()
````

````
female = this_data[this_data['sex']=='หญิง'] # แยกเป็นผู็หญิง
female['age'] # เลือกมาแค่อายุ
````

````
female['age'] = pd.to_numeric(female['age'],errors='coerce')
sum = 0
N = 0
for a in female['age']:

    if pd.notna(a) and a > 0: # เช็กว่าไม่ใช่ NaN และมากว่า 0
        sum += a # sum = sum + a
        N += 1

print(f'อายุเฉลี่ย ของ ผู้ป่วยหญิง {sum/N}')
````

````
female_KK = female[female['province_of_onset']=='ขอนแก่น']

sum = 0
N = 0
for a in female_KK['age']:

    if a > 0:
        sum += a # sum = sum + a
        N += 1

print(f'อายุเฉลี่ย ของ ผู้ป่วยหญิง ในจังหวัดขอนแก่น {sum/N}')
````

## Quiz ตัดตารางออกมาเฉพาะปี 2021 ```announce_date``` ในปี 2021

**Hint**
*   วนลูปหา index ของปี 2021
*   ตัดตารางมาเฉพาะ ปี 2021



````
# จุดมุ่งหมมายของเราคือจะตัดมาเฉพราะปี 2021 โดยเราจะดูจาก announce_date
# ซึ่งตอนนี้มันเป็น format เดือน/วัน/ปี อยู่ซึ่งเราอยากได้แค่ปี2021 อย่างเดียว
# สิ่งที่เราทำคือวนลูปเข้าไปในแถวแต่ละแถวพอวนลูปเสร็จ
# แต่ทีนี้เราอยากใช้ค่า announce_date แต่หน้าตามันยังเป็น format อยู๋เราอยากได้แค่ปี เลยต้อง split('/') เพราะ format มันแบ่งด้วย /
# เราก็จะได้เป็น เดือนคือตำแหน่งที่ 0 , วันตำแหน่งที่, ปีตำแหน่งที่ 2 และเราต้องการแค่ก็ indict ไปที่ 2 แล้วกำหนดปีไปก็คือ == '2021'
# พอเราได้ปีมา 2021 มาแล้วเราก็มาสร้าง list ก็คือถ้าเจอปี 2021 จะใส่ว่า True แต่ถ้าไม่ใช่ให้ใส่ false
# สุดท้าย print ออกมาถ้ามีชื่อตารางก็จะออกมาทั้งตารางเลยที่มี 2021 แต่ถ้า Print แค่ TF ก็จะออกมาเป็น True and False
TF=list()
for each_row in data_covid.iterrows():
    if each_row[1]['announce_date'].split('/')[2] == '2021':
        TF.append(True)
    else:
        TF.append(False)

data_covid[TF].head()
````
